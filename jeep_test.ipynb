```python
import sys
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# auto-reload library if developing library functionalities
%reload_ext autoreload
%autoreload 2
# Add the old-can-decoder-c56d4fe6 directory to sys.path
sys.path.append(os.path.abspath('old-can-decoder-c56d4fe6'))

# import library
from PandaCANDecoder.decoder import Decoder as OldDecoder
from can_decoder.decoder import Decoder
```

```python
decoder = Decoder(csv_file_path = "./data/SampleRoadAcceleration_v2.csv")

# Generate message objects

byte_filters = {
    "18DA10F1": ["1", "2", "3", "4",'5'],
    "18DAF110": ["1", "2", "3", "4",'5'],
    "18DAF218": ["1", "2", "3", "4",'5'],
    "18DA18F2": ["1", "2", "3", "4",'5'],
}

decoder.generate_msgs(byte_filters=byte_filters)
```

```python
## Dynamic filtering: if byte 1 is 0 we should discard byte 5
for msg in decoder.msgs:
    if msg.msg_byte_filter is not None:
        if msg.msg_byte_filter['1'] != 0:
            msg.msg_byte_filter.pop('5', None)
            # Remove duplicates

# Remove duplicate messages with same msg_id and identical msg_byte_filter after discarding byte 5
unique_msgs = []
seen = set()
for msg in decoder.msgs:
    key = (msg.msg_id, tuple(sorted(msg.msg_byte_filter.items())) if msg.msg_byte_filter else None)
    if key not in seen:
        unique_msgs.append(msg)
        seen.add(key)
decoder.msgs = unique_msgs
```

```python
# Generate message time series data

decoder.generate_msg_ts_data(rewrite=True)  
```

```python
decoder.calculate_signals(
        tokenization_method='conditional_bit_flip',
        signedness_method='msb_classifier',
        alpha1=0.01,
        alpha2=0.5, #0.5,
        gamma1=0.2)
```

## Engine RMP

```python
# Engine RPM
message_rpm = decoder.get_message('18DAF110', [5, 98, 1, 213])
decoder.plot_message_from_id('18DAF110', [5, 98, 1, 213])
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

alpha1_slider = widgets.FloatSlider(value=0.01, min=0.0, max=0.1, step=0.001, description='alpha1')
alpha2_slider = widgets.FloatSlider(value=0.5, min=0.0, max=1.0, step=0.01, description='alpha2')
realtime_checkbox = widgets.Checkbox(value=True, description='Real-time update')
update_button = widgets.Button(description='Update')
output = widgets.Output()

def update_plot(change=None):
    with output:
        clear_output(wait=True)
        decoder.calculate_signal(
            message_rpm,
            tokenization_method='conditional_bit_flip',
            signedness_method='msb_classifier',
            alpha1=alpha1_slider.value,
            alpha2=alpha2_slider.value,
            gamma1=0.2
        )
        message_rpm.plot()

def setup_observers():
    if realtime_checkbox.value:
        alpha1_slider.observe(update_plot, names='value')
        alpha2_slider.observe(update_plot, names='value')
        update_button.on_click(lambda x: None)  # Remove button action
    else:
        alpha1_slider.unobserve(update_plot, names='value')
        alpha2_slider.unobserve(update_plot, names='value')
        update_button.on_click(lambda x: update_plot())

def on_checkbox_change(change):
    setup_observers()

realtime_checkbox.observe(on_checkbox_change, names='value')
setup_observers()

# Initial display
update_plot()

display(widgets.VBox([alpha1_slider, alpha2_slider, realtime_checkbox, update_button, output]))
```
